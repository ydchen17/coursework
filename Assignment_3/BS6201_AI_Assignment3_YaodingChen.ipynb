```python
class lottery_numbers:
    def __init__(self):
        self.prize1 = 10
        self.prize2 = 50
        self.prize3 = 250
        self.prize0 = 690
    def chance(self):
        sum_num = self.prize1 + self.prize2 + self.prize3 + self.prize0
        self.prize1chance = self.prize1/sum_num
        self.prize2chance = self.prize2/sum_num
        self.prize3chance = self.prize3/sum_num
        self.prize0chance = self.prize0/sum_num    
    def declare(self):
        self.chance()
        print("the chance of getting first prize is "+str(self.prize1chance))
        print("the chance of getting second prize is "+str(self.prize2chance))
        print("the chance of getting third prize is "+str(self.prize3chance))
        print("the chance of getting no prize is "+str(self.prize0chance))
```

```python
class lottery_numbers_v2:
    def __init__(self, prize1, prize2, prize3, sum_num):
        self.prize1 = prize1
        self.prize2 = prize2
        self.prize3 = prize3
        self.prize0 = sum_num - prize1 - prize2 - prize3
        self.sum_num = sum_num
    def chance(self):
        self.prize1chance = float(self.prize1)/self.sum_num
        self.prize2chance = float(self.prize2)/self.sum_num
        self.prize3chance = float(self.prize3)/self.sum_num
        self.prize0chance = float(self.prize0)/self.sum_num
    def declare(self):
        self.chance()
        print("the chance of getting first prize is "+str('{:.1%}'.format(self.prize1chance)))
        print(
            "the chance of getting second prize is "+
            str('{:.1%}'.format(self.prize2chance))
             )
        print("the chance of getting third prize is "+str('{:.1%}'.format(self.prize3chance)))
        print("the chance of getting no prize is "+str('{:.1%}'.format(self.prize0chance)))
    def change(self):
        prize_rank = input("What prize would you like to modify? Enter 1/2/3/0.")
        target_num = input("How many people do you want for this prize? Enter any number.")
        if prize_rank == "1":
            self.sum_num = self.sum_num - self.prize1 + int(target_num)
            self.prize1 = target_num
        elif prize_rank == "2":
            self.sum_num = self.sum_num - self.prize2 + int(target_num)
            self.prize2 = target_num
        elif prize_rank == "3":
            self.sum_num = self.sum_num - self.prize3 + int(target_num)
            self.prize3 = target_num
        elif prize_rank == "0":
            self.sum_num = self.sum_num - self.prize0 + int(target_num)
            self.prize0 = target_num
        #print(self.prize1)
        notification = input("\nDo you want to know the current chance of getting prizes? \nEnter Y/N.")
        if notification == "Y":
            self.declare()
```

```python
round2 = lottery_numbers_v2(10,100,500, 1500)
round2.change()
```

```python
class lottery_numbers_v3:
    def __init__(self, prize1, prize2, prize3, sum_num):
        self.prize1 = prize1
        self.prize2 = prize2
        self.prize3 = prize3
        self.prize0 = sum_num - prize1 - prize2 - prize3
        self.sum_num = sum_num
    def chance(self):
        self.prize1chance = float(self.prize1)/self.sum_num
        self.prize2chance = float(self.prize2)/self.sum_num
        self.prize3chance = float(self.prize3)/self.sum_num
        self.prize0chance = float(self.prize0)/self.sum_num
    def declare(self):
        self.chance()
        print("the chance of getting first prize is "+str('{:.1%}'.format(self.prize1chance)))
        print(
            "the chance of getting second prize is "+
            str('{:.1%}'.format(self.prize2chance))
             )
        print("the chance of getting third prize is "+str('{:.1%}'.format(self.prize3chance)))
        print("the chance of getting no prize is "+str('{:.1%}'.format(self.prize0chance)))
    def change(self):
        prize_rank = input("What prize would you like to modify? Enter 1/2/3/0.")
        target_num = input("How many people do you want for this prize? Enter any number.")
        if prize_rank == "1":
            self.sum_num = self.sum_num - self.prize1 + int(target_num)
            self.prize1 = target_num
        elif prize_rank == "2":
            self.sum_num = self.sum_num - self.prize2 + int(target_num)
            self.prize2 = target_num
        elif prize_rank == "3":
            self.sum_num = self.sum_num - self.prize3 + int(target_num)
            self.prize3 = target_num
        elif prize_rank == "0":
            self.sum_num = self.sum_num - self.prize0 + int(target_num)
            self.prize0 = target_num
        #print(self.prize1)
        notification = input("\nDo you want to know the current chance of getting prizes? \nEnter Y/N.")
        if notification == "Y":
            self.declare()
    def predict(self):
        self.chance()
        target_prize = int(input("What prize would you like to get? Enter 1/2/3/0."))
        round_num = int(input("How many rounds do you want to play? Enter a number."))
        chances_once = [self.prize0chance, self.prize1chance, self.prize2chance, self.prize3chance]
        chance_once = 1 - (1-chances_once[target_prize])**round_num
        print("You have "+str('{:.1%}'.format(chance_once))+" chance to get the prize for once or more.")
```

```python
round2 = lottery_numbers_v3(10,100,500, 1500)
round2.declare()
```
