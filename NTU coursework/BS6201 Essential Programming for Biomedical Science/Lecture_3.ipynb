### Assignment 2: 
Build a rock-paper-scissor game

Assume that there are two players, and we're going to take their inputs

Players type either "rock", "paper", or "scissors"
- rock beats scissors
- scissors beats paper
- paper beats rock

To-do:
1. take inputs from user 1 and 2
2. compare inputs
3. print the winner

Bonus:
1. Keep the game going until either player doesn't want to play anymore
    - "stop"
    - "continue"
2. Hint: nested if/else, while


```python
u1 = input("Choose rock, paper, or scissors ")
u2 = input("Choose rock, paper, or scissors ")

if u1 == u2: 
    print("It's a tie!")
    
elif u1 == 'rock':
    if u2 == 'paper':
        print("User 2 wins!")
    elif u2 == 'scissors':
        print("User 1 wins!")
        
elif u1 == 'paper':
    if u2 == 'rock':
        print("User 1 wins!")
    elif u2 == 'scissors':
        print("User 2 wins!")
        
elif u1 == 'scissors':
    if u2 == 'paper':
        print("User 1 wins!")
    elif u2 == 'rock':
        print("User 2 wins!")
        
else:
    print("Please use one of the three choices!")
```

```python
# create a condition where user 1 wins
u1 = input("Choose rock, paper, or scissors ")
u2 = input("Choose rock, paper, or scissors ")

if (u1 == 'rock' and u2 == 'scissors') or (u1 == 'paper' and u2 == 'rock') or (u1 == 'scissors' and u2 == 'paper'):
    print("User 1 wins!")
    
elif (u2 == 'rock' and u1 == 'scissors') or (u2 == 'paper' and u1 == 'rock') or (u2 == 'scissors' and u1 == 'paper'):
    print("User 2 wins!")
```

```python
cont = "c"

while cont == "c":
    u1 = input("Choose rock, paper, or scissors ")
    u2 = input("Choose rock, paper, or scissors ")

    if (u1 == 'rock' and u2 == 'scissors') or (u1 == 'paper' and u2 == 'rock') or (u1 == 'scissors' and u2 == 'paper'):
        print("User 1 wins!")

    elif (u2 == 'rock' and u1 == 'scissors') or (u2 == 'paper' and u1 == 'rock') or (u2 == 'scissors' and u1 == 'paper'):
        print("User 2 wins!")
        
    cont = input("Enter c to keep playing")
    print(" ")
```

### Imagine you are building a house

### In OOP:
Think of class as the blueprint

Think of object as the things you work with

In an object, you have two things:
1. attributes
2. methods

If you have a ball, what are the attributes?
- size
- color
- price

What can you do with the ball? (methods)
- kick
- bounce
- throw
- pat

```python
# create the "blueprint" for a Ball

class Ball: # classes are capitalized
    # init method (init = initialize) is a method that runs the moment an object is created
    def __init__(self, name, shape, color, price): # self is the name for itself
        self.name = name # declare that the object has an attribute called name
        self.shape = shape
        self.color = color
        self.price = price
```

```python
ball_1 = Ball('Bob', 'round', 'black', 5)
```

```python
ball_1.shape
```

```python
ball_1.name # this is an attribute
```

```python
# add additional attributes
# object.some_attribute = some_value
ball_1.shape = 'round'
```

```python
ball_1.name = 'Bobobob'
ball_1.name
```

```python
class Person:
    def __init__(self, name):
        self.name = name
        self.wallet = 100 # amount of money I have
        self.quotes = ["Hello", "Good day", "Hi", "Woohoo"]
        
    # define method
    def say_hello(self):
        print("Hello world!") # standalone method
        
    def say_own_name(self):
        print("Hello, my name is " + self.name) # method that refers to an attribute
        
    def say_something(self, sentence):
        print("I would like to say: " + sentence) # method that takes an external argument
        
    def spend_money(self, spend_amount): # method that manipulates attributes
        self.wallet -= spend_amount # -= means decrease value in variable
        print("I have " + str(self.wallet) + " dollars left")

    def spend_money_v2(self, spend_amount): # method that involves control flow
        if (self.wallet - spend_amount) >= 0:
            self.wallet -= spend_amount # -= means decrease value in variable
            print("I have " + str(self.wallet) + " dollars left")
        else:
            print("Oh I don't have enough")
            
    def make_quote(self):
        print(rd.choice(self.quotes)) # selects an item randomly from a list
```

```python
jackie = Person("Jackie")
james = Person("James")
```

```python
jackie.make_quote()
```
