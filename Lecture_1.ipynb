```python
# create a function called "dog_years"
# takes in 1 parameter, age
# function will multiply age by 7
# return the dog age

# Level 1 - complete the function
# Level 2 - ask the user for how old the dog is, then return dog age

# Come back at 742pm
```

```python
# Level 1
def dog_years(age):
    dog_age = age * 7 # declare a variable first
    return "Your dog is " + str(dog_age) + " dog years old."

print(dog_years(3))
```

```python
# Level 2 - take user's input
user_dog_age = input("How long have you had your dog? ")
user_dog_age_int = int(user_dog_age)

def dog_years(age):
    dog_age = age * 7 # declare a variable first
    return "Your dog is " + str(dog_age) + " dog years old."

print(dog_years(user_dog_age_int))
```

```python
# Level 2 - take user's input (combine int and input)
user_dog_age_int = float(input("How long have you had your dog? ")) # wrap one function in another func
#user_dog_age_int = int(user_dog_age)

def dog_years(age):
    dog_age = age * 7 # declare a variable first
    return "Your dog is " + str(dog_age) + " dog years old."

print(dog_years(user_dog_age_int))
```
