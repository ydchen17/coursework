```python
# declare variable weather
weather = 'sunny'

if weather == 'raining':
    print("I will bring an umbrella")

# else is the default code to be run if above if doesn't run    
else:
    print("I will not bring an umbrella")
```

```python
# declare variable weather
weather = 'windy'

if weather == 'raining': # condition 1
    print("I will bring an umbrella")

elif weather == 'sunny': # condition 2, elif is else if
    print("I will also bring an umbrella")
    
# else is the default code to be run if above if doesn't run    
else:
    print("I will not bring an umbrella")
```

```python
# nested if/else statements
weather = "sunny"
raincoat = "yes"

# check if weather is raining, then check if I have raincoat
if weather == 'raining':
    if raincoat == 'yes':
        print("I will be fully protected")
    else:
        print("I will be slightly wet")
```

```python
def bmi_calculator(weight, height):
    return weight / height**2
```

```python
# < 19 = underweight
# 36 <= BMI < 40 is moderately obese
# > 54 = tell the user to see a doctor
# create a program called bmi_status
# takes user inputs
# tells user what he/she is based on BMI status

# Time: Come back at 633pm

def bmi_status(bmi):
    if bmi < 19:
        print("You're underweight")
    elif 19 <= bmi < 25:
        print("You're normal")
    elif 25 <= bmi < 30:
        print("You're overweight")
    elif 30 <= bmi < 35:
        print("You're obese")
    elif 35 <= bmi < 40:
        print("You're moderately obese")
    elif 40 <= bmi < 54:
        print("You're extremely obese")
    else:
        print("Go see a doctor")
        
height = float(input("What's your height in m? "))
weight = float(input("What's your weight in kg? "))

bmi_status(bmi_calculator(weight, height))
```
