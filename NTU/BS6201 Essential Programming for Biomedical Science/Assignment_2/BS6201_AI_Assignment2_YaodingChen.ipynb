# Demo

The idea here is quite simple. We have two players. Player 1 is known to be human but we might not be so sure about Player 2. Therefore, we will ask about whether Player 2 is a computer first. Then, if it is a computer, it will randomly give an input for comparison. When human players play, their input will be checked, and if not in the known input list, they will be asked whether they want to continue to play until they say stop.

```python
from random import randint

#create a list of play options
t = ["Rock", "Paper", "Scissors"]

#set player1 to False
player1 = False
player2_type = input("Do you want to play with computer? Yes or No?")

while player1 == False:
    
    #assign a random play to the player2
    player1 = input("Rock, Paper, Scissors?")
    if player1 not in t:
        con1 = input("That's not a valid play. Stop or continue?")
        if con1 == "Stop" or con1 == "stop":
            break
    if player2_type == "Yes":
        player2 = t[randint(0,2)]
    else: 
        player2 = input("Rock, Paper, Scissors?")
        if player2 not in t:
            con2 = input("That's not a valid play. Stop or continue?")
            if con2 == "Stop" or con2 == "stop":
                break
    if player1 == player2:
        print("Tie!")
    elif player1 == "Rock":
        if player2 == "Paper":
            print("Player 2 wins!", player2, "covers", player1)
        else:
            print("Player 1 wins!", player1, "smashes", player2)
    elif player1 == "Paper":
        if player2 == "Scissors":
            print("Player 2 wins!", player2, "cut", player1)
        else:
            print("Player 1 wins!", player1, "covers", player2)
    elif player1 == "Scissors":
        if player2 == "Rock":
            print("Player 2 wins!", player2, "smashes", player1)
        else:
            print("Player 1 wins!", player1, "cut", player2)
            
    player1 = False
    player2 = t[randint(0,2)]
```
