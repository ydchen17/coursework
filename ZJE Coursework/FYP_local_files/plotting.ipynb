## 2. Olink panel-based enrichment

```r
ORCADES_olink <- read.csv("ORCADES-olink.csv")
ORCADES_olink
```

```r
# odds for matched = 51 / (31492 - 51)
# odds for human-inserted = 5 / (2471-5)
# odds ratio = (5/2466)/(51/31441)
# In R, fisher.test(matrix(c(5,2466,51,31441),ncol=2))
```

```r
ORCADES_olink <- read.csv("ORCADES-olink.csv")

ORCADES_olink_fisher = ORCADES_olink
# amb, exp, hum, m-del, m-dim, mat, shu
total = c(14400, 8782, 2472, 11249, 8051, 76445, 31491)

fisher_results_list = vector()
#headers <- c("cm", "cr", "cvd2", "cvd3", "dev", "imr", "inf1", "met", "neurexp", "neuro", "oncii", "ordam")
headers <- c("cardiometabolic", "cell regulation", "CVD II", "CVD III", "development", "immune response", "inflammation", "metabolism", "neuro exploratory", "neurology", "oncology II", "organ damage")

for (i in 1:(length(ORCADES_olink_fisher$X)-1)){
    for (j in 2:length(ORCADES_olink_fisher[i,])){
        fisher = fisher.test(matrix(c(ORCADES_olink_fisher[i,j], 
                                            total[i]-ORCADES_olink_fisher[i,j],
                                            ORCADES_olink_fisher[7,j], 
                                            total[7]-ORCADES_olink_fisher[7,j]),
                                          ncol=2))
        fisher_results = c(
            evo_group = ORCADES_olink_fisher[i,1], olink_group = headers[j-1], 
            count = ORCADES_olink_fisher[i,j],
            odds_ratio = as.numeric(fisher$estimate), 
            lower_conf.int = as.numeric(fisher$conf.int[1]), 
            higher_conf.int = as.numeric(fisher$conf.int[2]), 
            p.value = as.numeric(fisher$p.value))
        fisher_results_list = rbind(fisher_results_list, fisher_results)
    }
}

fisher_results_list = as.data.frame(fisher_results_list)
# V1 = evo group,  V2 = olink panel, V3 = Odds Ratio, V4= lower conf.int, V5 = higher conf.int, V6= p-value
```

```r
write.csv(fisher_results_list, "ORCADES_fisher_results_list.csv")
```

```r
# install.packages("dplyr")
library(dplyr)
fisher_results_significant <- filter(fisher_results_list, p.value < 0.05)
fisher_results_significant
```

```r
library(ggplot2)
p = ggplot(fisher_results_significant,aes(evo_group,olink_group))
p=p + geom_point()  
# 修稿点的大小
p=p + geom_point(aes(size=as.numeric(count)))
# 展示三维数据
pbubble = p+ geom_point(aes(size=as.numeric(count),color=as.numeric(odds_ratio)))
# 设置渐变色
pr = pbubble+scale_color_gradient(low="blue",high = "red")
# 绘制p气泡图
pr = pr+labs(color="Odds ratio",
             size="Count",  
             x="Evolutionary history profiles",
             y="Olink panels",
             title="Olink panel enrichment") 

pr = pr + theme_bw() + 
    theme(axis.text.x = element_text(angle = 30, hjust = 1, vjust = 1)) + 
    scale_size_continuous(range = c(1, 10))+
    theme(text = element_text(size=20))
print(pr)
## 保存图片
#ggsave("out.pdf")# 保存为pdf格式
#ggsave("out2.png",width=4,height=4)# 设定画布大小
```
