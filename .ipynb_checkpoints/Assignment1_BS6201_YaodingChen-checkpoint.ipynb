Create a function bmi_calculator that takes in your height and weight to calculate your body's BMI. The function takes in two parameters, and returns the user of the function the BMI. ## Demo 1 

```python
# define a function named bmi_calculator with two inputs, namely weight, height.
def bmi_calculator(weight, height):   
    bmi = weight/(height**2)       # BMI = weight / height^2
    return bmi                     # the function returns BMI
```

```python
bmi_calculator(1.77,80)            # test run 
```

Let's make it more interactive with input()

```python
weight_x = float(input("what is your weight (kg)?"))
```

```python
height_x = float(input("what is your height (m)?"))
```

Therefore, the result of BMI calculation equates

```python
bmi_calculator(weight_x,height_x)
```

## Demo 2

Let's put all things together in one function, so we can re-write a bmi_calculator function as follows:

```python
def bmi_calculator():
    weight = float(input("what is your weight (kg)?"))
    height = float(input("what is your height (m)?"))
    bmi = weight/(height**2)
    print("Your BMI equates "+str(bmi))
    return bmi
```

Below is a test run result.

```python
bmi_calculator()
```
